# Silver Layer – Data Transformation

## Purpose

Transform raw Bronze data into cleansed, standardised and validated datasets while preserving the business meaning of the original CRM and ERP sources.

The Silver layer addresses data quality issues, standardises attributes and applies the required transformation rules to make the data consistent and suitable for downstream analytical processing.

## Source Systems

- CRM: Customer, product and sales information.
- ERP: Customer, location and product category information.

```sql
CREATE TABLE IF NOT EXISTS sales.silver.crm_cust_info (
    cst_id INT,
    cst_key STRING,
    cst_firstname STRING,
    cst_lastname STRING,
    cst_marital_status STRING,
    cst_gndr STRING,
    cst_create_date DATE,
    dwh_create_date TIMESTAMP DEFAULT CURRENT_TIMESTAMP()
)
USING DELTA
TBLPROPERTIES ('delta.feature.allowColumnDefaults' = 'supported')
COMMENT 'Customer records extracted from the CRM system.';
```

```sql
CREATE TABLE IF NOT EXISTS sales.silver.crm_prd_info (
    prd_id INT,
    prd_key STRING,
    prd_nm STRING,
    prd_cost INT,
    prd_line STRING,
    prd_start_dt DATE,
    prd_end_dt DATE,
    dwh_create_date TIMESTAMP DEFAULT CURRENT_TIMESTAMP()
)
USING DELTA
TBLPROPERTIES ('delta.feature.allowColumnDefaults' = 'supported')
COMMENT 'Product records extracted from the CRM system.';
```

```sql
CREATE TABLE IF NOT EXISTS sales.silver.crm_sales_details (
    sls_ord_num STRING,
    sls_prd_key STRING,
    sls_cust_id INT,
    sls_order_dt INT,
    sls_ship_dt INT,
    sls_due_dt INT,
    sls_sales INT,
    sls_quantity INT,
    sls_price INT,
    dwh_create_date TIMESTAMP DEFAULT CURRENT_TIMESTAMP()
)
USING DELTA
TBLPROPERTIES ('delta.feature.allowColumnDefaults' = 'supported')
COMMENT 'Sales transaction records extracted from the CRM system.';
```

```sql
CREATE TABLE IF NOT EXISTS sales.silver.erp_loc_a101 (
    cid STRING,
    cntry STRING,
    dwh_create_date TIMESTAMP DEFAULT CURRENT_TIMESTAMP()
)
USING DELTA
TBLPROPERTIES ('delta.feature.allowColumnDefaults' = 'supported')
COMMENT 'Location and country information from the ERP system.';
```

```sql
CREATE TABLE IF NOT EXISTS sales.silver.erp_cust_az12 (
    cid STRING,
    bdate DATE,
    gen STRING,
    dwh_create_date TIMESTAMP DEFAULT CURRENT_TIMESTAMP()
)
USING DELTA
TBLPROPERTIES ('delta.feature.allowColumnDefaults' = 'supported')
COMMENT 'Customer birth date and gender information from the ERP system.';
```

```sql
CREATE TABLE IF NOT EXISTS sales.silver.erp_px_cat_g1v2 (
    id STRING,
    cat STRING,
    subcat STRING,
    maintenance STRING,
    dwh_create_date TIMESTAMP DEFAULT CURRENT_TIMESTAMP()
)
USING DELTA
TBLPROPERTIES ('delta.feature.allowColumnDefaults' = 'supported')
COMMENT 'Product category, subcategory and maintenance information from the ERP system.';
```
