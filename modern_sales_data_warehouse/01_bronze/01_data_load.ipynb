# Bronze Layer – Data Load

Load the CRM and ERP source files into their corresponding Bronze Delta tables.

COPY INTO is used to load the files while maintaining file-level ingestion tracking. Previously loaded files are not reprocessed during subsequent executions.

```sql
-- CRM customer
COPY INTO sales.bronze.crm_cust_info
FROM '/Volumes/sales/source/raw/source_crm'
FILEFORMAT = CSV
FILES = ('cust_info.csv')
FORMAT_OPTIONS (
    'header' = 'true',
    'inferSchema' = 'true'
);
```
