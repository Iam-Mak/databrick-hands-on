# Bronze Layer – Table Setup

## Purpose

Create the Bronze Delta tables for data received from the CRM and ERP source systems.

The tables retain the source structure and data types as closely as possible. Data cleansing, standardisation and business transformations will be handled in the Silver layer.

## Source Systems

- CRM: Customer, product and sales information.
- ERP: Customer, location and product category information.

```sql
CREATE TABLE IF NOT EXISTS sales.bronze.crm_cust_info (
    cst_id INT,
    cst_key STRING,
    cst_firstname STRING,
    cst_lastname STRING,
    cst_marital_status STRING,
    cst_gndr STRING,
    cst_create_date DATE
)
USING DELTA
COMMENT 'Customer records extracted from the CRM system.';
```

```sql
CREATE TABLE IF NOT EXISTS sales.bronze.crm_prd_info (
    prd_id INT,
    prd_key STRING,
    prd_nm STRING,
    prd_cost INT,
    prd_line STRING,
    prd_start_dt DATE,
    prd_end_dt DATE
)
USING DELTA
COMMENT 'Product records extracted from the CRM system.';
```

```sql
CREATE TABLE IF NOT EXISTS sales.bronze.crm_sales_details (
    sls_ord_num STRING,
    sls_prd_key STRING,
    sls_cust_id INT,
    sls_order_dt INT,
    sls_ship_dt INT,
    sls_due_dt INT,
    sls_sales INT,
    sls_quantity INT,
    sls_price INT
)
USING DELTA
COMMENT 'Sales transaction records extracted from the CRM system.';
```

```sql
CREATE TABLE IF NOT EXISTS sales.bronze.erp_loc_a101 (
    cid STRING,
    cntry STRING
)
USING DELTA
COMMENT 'Location and country information from the ERP system.';
```

```sql
CREATE TABLE IF NOT EXISTS sales.bronze.erp_cust_az12 (
    cid STRING,
    bdate DATE,
    gen STRING
)
USING DELTA
COMMENT 'Customer birth date and gender information from the ERP system.';
```

```sql
CREATE TABLE IF NOT EXISTS sales.bronze.erp_px_cat_g1v2 (
    id STRING,
    cat STRING,
    subcat STRING,
    maintenance STRING
)
USING DELTA
COMMENT 'Product category, subcategory and maintenance information from the ERP system.';
```
