# Sales Data Warehouse – Initial Setup

## Project Overview

This project implements an end-to-end Sales Data Warehouse using Databricks. It integrates data from CRM and ERP source systems and organises it into a medallion architecture to support reliable data transformation and analytical reporting.

## Data Architecture

The warehouse follows three processing layers:

- **Bronze:** Raw data ingested from CRM and ERP source files.
- **Silver:** Cleansed, standardised and integrated data.
- **Gold:** Business-ready data models designed for analytical consumption.



```sql
CREATE CATALOG IF NOT EXISTS sales
COMMENT 'Sales data warehouse built using CRM and ERP source data. Contains customer, product and sales data processed through Bronze, Silver and Gold layers.';
```

```sql
CREATE SCHEMA IF NOT EXISTS sales.source
COMMENT 'Schema for source file storage and source related objects.';

CREATE SCHEMA IF NOT EXISTS sales.bronze
COMMENT 'Raw tables containing data ingested from CRM and ERP source files.';

CREATE SCHEMA IF NOT EXISTS sales.silver
COMMENT 'Cleaned and standardised data, with CRM and ERP records integrated where required.';

CREATE SCHEMA IF NOT EXISTS sales.gold
COMMENT 'Business level tables for sales, customer and product analysis.';
```

```sql
CREATE VOLUME IF NOT EXISTS sales.source.raw
COMMENT 'Stores the original CRM and ERP CSV files before ingestion into the Bronze layer.';
```
